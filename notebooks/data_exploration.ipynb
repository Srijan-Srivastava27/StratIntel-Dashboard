```python
# Cell 1: Setup
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add src to path
sys.path.append('../src')

from data_collector import DataCollector
from sentiment_analyzer import SentimentAnalyzer

%matplotlib inline
plt.style.use('seaborn-v0_8')
```

```python
# Cell 2: Load and explore data
collector = DataCollector()
queries = ["AI trends", "market analysis", "technology news"]

# Collect sample data
sample_data = collector.collect_all_data(queries)
print(f"Collected {len(sample_data)} items")
sample_data.head()
```

```python

# Cell 3: Data visualization
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Source distribution
sample_data['source'].value_counts().plot(kind='bar', ax=axes[0,0])
axes[0,0].set_title('Data Sources Distribution')

# Timeline analysis (if created_at exists)
if 'created_at' in sample_data.columns:
    sample_data['created_at'] = pd.to_datetime(sample_data['created_at'])
    sample_data.set_index('created_at').resample('H').size().plot(ax=axes[0,1])
    axes[0,1].set_title('Data Collection Over Time')

# Text length distribution
if 'text' in sample_data.columns:
    sample_data['text_length'] = sample_data['text'].str.len()
    sample_data['text_length'].hist(bins=50, ax=axes[1,0])
    axes[1,0].set_title('Text Length Distribution')

plt.tight_layout()
plt.show()

```

```python
# Cell 4: Sentiment analysis
analyzer = SentimentAnalyzer()

# Analyze sentiment for sample texts
sample_texts = sample_data['text'].fillna('').head(5).tolist()
sentiment_results = analyzer.batch_analyze(sample_texts)

# Display results
for i, result in enumerate(sentiment_results):
    print(f"\nText {i+1}: {sample_texts[i][:100]}...")
    print(f"Sentiment: {result['sentiment']} (Confidence: {result['confidence']:.2f})")
    print(f"Market Relevance: {result['market_relevance']}")
```
